```python
ratings_df = pd.read_csv("C:/Users/y/Desktop/Academics/projects/machinelearning/book_recommendation_system/datasets/BX-Book-Ratings.csv", sep=";", on_bad_lines="skip", encoding="latin-1")
```

```python
ratings_df.head(2)
```

```python
ratings_df.rename(columns={
    'User-ID': 'user_id',
    'ISBN': 'isbn',
    'Book-Rating': 'rating'
}, inplace=True)
```

```python
ratings_df.head(2)
```

```python
ratings_df.shape
```

```python
user_book_counts = ratings_df['user_id'].value_counts()
heavy_readers = user_book_counts[user_book_counts > 200].index
ratings_df = ratings_df[ratings_df['user_id'].isin(heavy_readers)]
```

```python
ratings_df.head(2)
```

```python
rating_with_books = ratings_df.merge(books_df, on='isbn')
```

```python
rating_with_books.head(2)
```

```python
number_rating  = rating_with_books.groupby('title')['rating'].count().reset_index()
```

```python
number_rating.rename(columns={'no_od_ratings': 'no_of_ratings'}, inplace=True)
```

```python
number_rating.head(2)
```

```python
final_rating = rating_with_books.merge(number_rating, on='title')
final_rating.head(2)
```

```python
final_rating_filtered = final_rating[final_rating['no_of_ratings'] > 50]
```

```python
final_rating_filtered.shape
```

```python
final_rating_filtered.drop_duplicates(['user_id', 'title'], inplace=True)
```

```python
final_rating_filtered.head(2)
```

```python
book_pivot = final_rating_filtered.pivot_table(columns='user_id', index='title', values='rating')
```

```python
book_pivot.fillna(0, inplace=True)
```

```python
book_pivot
```

```python
from scipy.sparse import csr_matrix
```

```python
book_sparse = csr_matrix(book_pivot)
```

```python
book_sparse
```

```python
from sklearn.neighbors import NearestNeighbors
model = NearestNeighbors(algorithm='brute')
```

```python
model.fit(book_sparse)
```

```python
book_pivot.iloc[237, :]
```

```python
distance, suggestion = model.kneighbors(book_pivot.iloc[237, :].values.reshape(1, -1), n_neighbors=6)
```

```python
# Testing the Model Suggestion
def recommend_book(book_name):
    book_id = np.where(book_pivot.index == book_name)[0][0]
    distance, suggestion = model.kneighbors(book_pivot.iloc[book_id,:].values.reshape(1, -1), n_neighbors=6)

    for i in range(len(suggestion)):
        books = book_pivot.index[suggestion[i]]
        for j in books:
            if j == book_name:
                print(f"You searched: {book_name}\n")
                print(f"The suggestion boooks are: \n")
            else:
                print(j)
```

```python
recommend_book('Hemlock Bay')
```
